```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv("creditcard.csv")
```

```python
df["Class"].value_counts()
#imbalanced dataset
```

upsampling

```python
df_minority = df[df["Class"] == 1]
```

```python
df_majority = df[df["Class"] == 0]
```

```python
df_minority
```

```python
from sklearn.utils import resample
```

```python
df_minority_upsampled = resample(df_minority, replace = True, n_samples = len(df_majority), random_state = 42)
```

```python
df_minority_upsampled
```

```python
df_upsampled = pd.concat([df_majority, df_minority_upsampled])
```

```python
df_upsampled["Class"].value_counts()
```

```python
plt.scatter(df_upsampled["V1"], df_upsampled["V2"])
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV
```

```python
X = df_upsampled.drop("Class", axis = 1)
y = df_upsampled["Class"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 15)
```

```python
rf = RandomForestClassifier()
```

```python
rf_params = {
    "estimator": [100,200,300,600],
    "max_depth": [5,9,16, None],
    "min_samples_split":[2,4,9,16],
    "min_samples_leaf":[1,2,3,4]  
    }
```

```python
rfc = RandomizedSearchCV(estimator = rf, param_distributions=rf_params, cv=3, n_jobs=-1)
rfc.fit(X_train, y_train)
```

```python
y_pred = rfc.predict(X_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV

# First, create the base model
rf = RandomForestClassifier()

# Define the parameter grid
rf_params = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 5, 10],
    'min_samples_split': [2, 4, 6],
    'min_samples_leaf': [1, 2, 4]
    # Add other parameters as needed
}

# Create the RandomizedSearchCV object
rfc = RandomizedSearchCV(estimator=rf, param_distributions=rf_params, cv=3, n_jobs=-1)

# Fit the model
rfc.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
print(accuracy_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_pred, y_test))
```
